# 🔍 Satellite Chip Data Verification
업로드된 .npy 및 .json 파일을 확인하고 시각화합니다.

```python
import numpy as np
import json
import matplotlib.pyplot as plt

# 1. 데이터 로드
data = np.load('2015-11-02.npy')
with open('2015-11-02_meta.json', 'r') as f:
    meta = json.load(f)

print(f'Metadata: {meta}')
print(f'Data Shape: {data.shape} (Channels, Height, Width)')
```

```python
# 2. 채널별 시각화
fig, axes = plt.subplots(1, 5, figsize=(20, 4))
for i in range(5):
    axes[i].imshow(data[i], cmap='viridis')
    axes[i].set_title(f'Channel {i}')
    axes[i].axis('off')
plt.tight_layout()
plt.show()
```
