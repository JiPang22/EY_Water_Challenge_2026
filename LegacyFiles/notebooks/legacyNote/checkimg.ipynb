```python
import os
import shutil
from glob import glob

# 1. 대상 폴더 정의
folders = [
    'notebooks', 
    'rawData/features', 
    'rawData/templates', 
    'models/weights', 
    'models/tunings'
]

for folder in folders:
    os.makedirs(folder, exist_ok=True)

# 2. 파일 이동 로직 (노트북 셀에서 실행 가능)
# 노트북 파일 이동
for nb in glob('*.ipynb'):
    if nb != 'Untitled.ipynb': # 작업 중인 파일 제외
        shutil.move(nb, f'notebooks/{nb}')

# 특성 CSV 파일 이동
for csv_feat in glob('*_features_*.csv'):
    shutil.move(csv_feat, f'rawData/features/{csv_feat}')

# 원본 및 템플릿 이동
if os.path.exists('water_quality_training_dataset.csv'):
    shutil.move('water_quality_training_dataset.csv', 'rawData/water_quality_training_dataset.csv')
if os.path.exists('submission_template.csv'):
    shutil.move('submission_template.csv', 'rawData/templates/submission_template.csv')

print("✔ 주피터 환경 파일 재구성 완료")
```
